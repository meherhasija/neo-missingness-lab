```python
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()

CSV_PATH = "/content/neo_asteroid_dataset.csv"
df = pd.read_csv(CSV_PATH)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nPHA counts:")
print(df["pha"].value_counts(dropna=False))

print("\nMissing values:")
print(df.isna().sum().sort_values(ascending=False))

display(df.head())
```

```python
FEATURES = ["a", "e", "i", "om", "w", "diameter"]
TARGET = "pha"

missing_columns = [
    col for col in FEATURES + [TARGET]
    if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

data = df.dropna(subset=[TARGET]).copy()

data[TARGET] = (
    data[TARGET]
    .astype(str)
    .str.strip()
    .str.upper()
    .map({"Y": 1, "N": 0})
)

data = data.dropna(subset=[TARGET]).copy()
data[TARGET] = data[TARGET].astype(int)

for col in FEATURES:
    data[col] = pd.to_numeric(data[col], errors="coerce")

X = data[FEATURES]
y = data[TARGET]

print("Prepared dataset:", data.shape)
print("\nTarget distribution:")
print(y.value_counts())

print("\nMissing values in features:")
print(X.isna().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
import sklearn
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.pipeline import Pipeline

print("Scikit-learn version:", sklearn.__version__)

model = Pipeline([
    ("classifier", HistGradientBoostingClassifier(
        max_iter=100,
        learning_rate=0.1,
        random_state=42,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

print("\nMethod 1 model trained successfully!")
print("Missing-value imputation: None")
print("Native NaN handling: Yes")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

results = {
    "Method": "Native Missing-Value Handling",
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred, zero_division=0),
    "Recall": recall_score(y_test, y_pred, zero_division=0),
    "F1-score": f1_score(y_test, y_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, y_prob),
    "Average Precision": average_precision_score(y_test, y_prob)
}

for metric, value in results.items():
    if isinstance(value, (int, float, np.integer, np.floating)):
        print(f"{metric}: {value:.4f}")
    else:
        print(f"{metric}: {value}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Non-PHA", "PHA"],
    zero_division=0
))
```
